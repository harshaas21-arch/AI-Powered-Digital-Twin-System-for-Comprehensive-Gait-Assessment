## Load Dataset


```python
import numpy as np
import os

# Dataset folder
DATASET_ROOT = "/content/dataset/UCI-HAR Dataset"

TRAIN_FOLDER = os.path.join(DATASET_ROOT, "train")
TEST_FOLDER = os.path.join(DATASET_ROOT, "test")

# Load data
X_train = np.loadtxt(os.path.join(TRAIN_FOLDER, "X_train.txt"))
y_train = np.loadtxt(os.path.join(TRAIN_FOLDER, "y_train.txt"))

X_test = np.loadtxt(os.path.join(TEST_FOLDER, "X_test.txt"))
y_test = np.loadtxt(os.path.join(TEST_FOLDER, "y_test.txt"))

print("Training Data :", X_train.shape)
print("Training Labels :", y_train.shape)

print("Testing Data :", X_test.shape)
print("Testing Labels :", y_test.shape)
```

```python
# Convert labels from 1-6 to 0-5

y_train = y_train.astype(int) - 1
y_test = y_test.astype(int) - 1

print("Unique Training Labels:", np.unique(y_train))
print("Unique Testing Labels:", np.unique(y_test))
```

## Building Neural Network

```python
from tensorflow import keras

model = keras.Sequential([
    keras.layers.Input(shape=(561,)),

    keras.layers.Dense(128, activation="relu"),

    keras.layers.Dense(64, activation="relu"),

    keras.layers.Dense(6, activation="softmax")
])

model.summary()
```

## Compiling the Model

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("✅ Model compiled successfully!")
```

## Training the Model

```python
history = model.fit(
    X_train,
    y_train,
    epochs=20,
    batch_size=32,
    validation_data=(X_test, y_test),
    verbose=1
)
```

## Saving the trained model

```python
import os

# Create models folder
os.makedirs("models", exist_ok=True)

# Save model
model.save("models/baseline_model.keras")

print("✅ Model saved successfully!")
```

```python
print(os.listdir("models"))
```

## Converting to TensorFlow Lite

```python
# Create converter
converter = tf.lite.TFLiteConverter.from_keras_model(model)

# Convert model
tflite_model = converter.convert()

# Save model
with open("models/baseline_model.tflite", "wb") as f:
    f.write(tflite_model)

print("✅ TensorFlow Lite model created!")
```

```python
import os

keras_size = os.path.getsize("models/baseline_model.keras") / 1024
tflite_size = os.path.getsize("models/baseline_model.tflite") / 1024

print(f"Keras Model Size : {keras_size:.2f} KB")
print(f"TFLite Model Size: {tflite_size:.2f} KB")
```

## Verifying TensorFlow Lite Model

```python
# Load TFLite model
interpreter = tf.lite.Interpreter(
    model_path="models/baseline_model.tflite"
)

interpreter.allocate_tensors()

# Input / Output information
input_details = interpreter.get_input_details()
output_details = interpreter.get_output_details()

print(input_details)
print(output_details)
```

## Comparing Predictions

```python
# Select one sample

sample = X_test[0:1]

# TensorFlow prediction

tf_prediction = model.predict(sample, verbose=0)

# TensorFlow Lite prediction

interpreter.set_tensor(
    input_details[0]['index'],
    sample.astype(np.float32)
)

interpreter.invoke()

tflite_prediction = interpreter.get_tensor(
    output_details[0]['index']
)

print("TensorFlow Prediction")
print(tf_prediction)

print("\nTensorFlow Lite Prediction")
print(tflite_prediction)
```

```python
print("\nTensorFlow Predicted Class :", np.argmax(tf_prediction))
print("TensorFlow Lite Predicted Class :", np.argmax(tflite_prediction))

print("Actual Label :", y_test[0])
```
