## Load Dataset


```python
import numpy as np
import os

# Dataset folder
DATASET_ROOT = "/content/dataset/UCI-HAR Dataset"

TRAIN_FOLDER = os.path.join(DATASET_ROOT, "train")
TEST_FOLDER = os.path.join(DATASET_ROOT, "test")

# Load data
X_train = np.loadtxt(os.path.join(TRAIN_FOLDER, "X_train.txt"))
y_train = np.loadtxt(os.path.join(TRAIN_FOLDER, "y_train.txt"))

X_test = np.loadtxt(os.path.join(TEST_FOLDER, "X_test.txt"))
y_test = np.loadtxt(os.path.join(TEST_FOLDER, "y_test.txt"))

print("Training Data :", X_train.shape)
print("Training Labels :", y_train.shape)

print("Testing Data :", X_test.shape)
print("Testing Labels :", y_test.shape)
```

```python
# Convert labels from 1-6 to 0-5

y_train = y_train.astype(int) - 1
y_test = y_test.astype(int) - 1

print("Unique Training Labels:", np.unique(y_train))
print("Unique Testing Labels:", np.unique(y_test))
```

## Building Neural Network

```python
from tensorflow import keras

model = keras.Sequential([
    keras.layers.Input(shape=(561,)),

    keras.layers.Dense(128, activation="relu"),

    keras.layers.Dense(64, activation="relu"),

    keras.layers.Dense(6, activation="softmax")
])

model.summary()
```

## Compiling the Model

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("✅ Model compiled successfully!")
```

## Training the Model

```python
history = model.fit(
    X_train,
    y_train,
    epochs=20,
    batch_size=32,
    validation_data=(X_test, y_test),
    verbose=1
)
```
